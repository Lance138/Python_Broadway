```python
#adding comma and more items makes it a tuple
t = (1,3)
r = (1)
s = ("Apple")
q = ("apple", "ball", "cat", "apple")
print(type(t))
print(type(r))
print(type(s))
print(type(q))
```

```python
q = ("apple", "ball", "cat", "apple")
print(q[0])
print(q[0:3])
print(q[0:5:2])
```

```python
#we can concat and change/update forcefully
a = (1,2,3,4,5)
b = ("Apple", "Ball")
c = a + b
print(c)
```

```python
#WAP a program to create a tuple
#This error shows tuple can only concat with another tuple
t = tuple()
n = int(input("Enter n = "))
for i in range(n):
    x = int(input("Enter x = "))
    t = t + x
print(t)
```

```python
#WAP a program to create a tuple
t = tuple()
n = int(input("Enter n = "))
for i in range(n):
    x = int(input("Enter x = "))
    t = t + (x,) #Here we put x inside a bracket and add a comma to convert the input from above code to tuple so that we can concat with t
print(t)
```

```python
#converting to tuple
a = ('apple', 'ball', 'apple', 'ball')
a = tuple(list(a))
a
```

```python
#We can search, find indexing in tuple
#We cannot remove in tuple
```

```python
#deleting item before converting to tuple
a = ('apple', 'ball', 'apple', 'ball')
a = list(a)
del a[0]
a = tuple(a)
a
```

```python
# tuple inside tuple
a = ((1,2,3),(4,5,6),(7,8,9))
# dict inside tuple (since this is a dict, we cannot delete the whole data but we can the data inside the dict)
b = ({'name':'Ram'},{'name':'Sita'})
# list inside tuple (we can change the items in the list but cannot change the entire tuple. We cannot add or completely delete
c = ([1,2,3],[4,5,6],[7,8,9])
print (a)
print (b)
print (c)
```
