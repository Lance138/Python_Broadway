```python
# Inheritance
class Data:
    def __init__(self):
        self.name = input("Enter name = " )
        self.age = int(input("Enter age = " ))
        self.add = input("Enter address = " )
    def data(self):
        print(self.name, self.age, self.add)

class Info(Data):
    def info(self):
        print(f"Hello World I am {self.name}. I am from {self.add} and I am {self.age} years old.")

obj = Info()
obj.info()
```

```python
#
class Data:
    def __init__(self, name, age, add):
        self.name = name
        self.age = age
        self.add = add
    def data(self):
        print(self.name, self.age, self.add)

# here data class is inherited by Info class. Since Info does not define its own __init__, it automatically uses the __init__ method from its parent class Data.”
class Info(Data):
    def info(self):
        print(f"Hello World I am {self.name}. I am from {self.add} and I am {self.age} years old.")

name = "Ram"
age = 23
add = "Kathmandu"
obj = Info(name, age, add)
obj.info()
```

```python
# Multilevel
class Name:
    def __init__(self):
        self.name = "Ram"
    def my_name(self):
        print(f"Hello World, my name is {self.name}")
        

class Age(Name):
    def __init__(self):
        self.age = 23
        Name.__init__(self)
    def my_age(self):
        print(f"Hello World, I am {self.age} years old!")

class Add(Age):
    def __init__(self):
        self.add = "Kathmandu"
        Age.__init__(self) #This tells the obj to go to the inherited class and call the __init__ function. Same thing for above
    def info(self):
        print(f"Hello World I am {self.name}. I am from {self.add} and I am {self.age} years old.")

obj = Add()
obj.info()
```

```python
# Multiple Inheritance
class Name:
    def __init__(self):
        self.name = "Ram"
    def my_name(self):
        print(f"Hello World, my name is {self.name}")
        

class Age:
    def __init__(self):
        self.age = 23
    def my_age(self):
        print(f"Hello World, I am {self.age} years old!")

class Add(Name, Age):
    def __init__(self):
        self.add = "Kathmandu"
        Name.__init__(self)
        Age.__init__(self) # Manually calling parent class constructors to initialize attributes from both Name and Age
    def info(self):
        print(f"Hello World I am {self.name}. I am from {self.add} and I am {self.age} years old.")

obj = Add()
obj.info()
obj.my_age()
obj.my_name()
```

```python
# Public attributes
class Office:
    #constructor
    def __init__(self):
        self.name = "Ram"
        self.sal = 13000
emp = Office()
emp.sal
```

```python
# Protected Attributes
class Office:
    #constructor
    def __init__(self):
        self._name = "Ram"
        self._sal = 13000

emp = Office()
emp._sal      # Need to call in protected method
```
