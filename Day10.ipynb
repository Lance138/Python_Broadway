```python
#CRUD
data = []
n = int(input("Enter no. of rows = "))
for i in range(n):
    name = input("Enter name = ")
    age = int(input("Enter age = "))
    address = input("Enter address = ")
    info = [name, age, address]
    data.append(info)

print(data)
```
