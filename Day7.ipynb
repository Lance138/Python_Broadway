```python
total2 = 0
bill = "" #empty to update later
n = int(input("Enter n = "))
for i in range (n):
    product = input("Enter product name = ")
    price = float(input("Enter price = "))
    qty = float(input("Enter quantity = "))
    total = price * qty
    total2 += total
    data = f"{product} {price} {qty} {total}\n"
    bill = bill + data #updating the data
    total2 = total + total2

print(bill)
print(total2)
```
