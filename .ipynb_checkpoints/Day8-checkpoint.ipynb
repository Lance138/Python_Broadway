```python
# append(), insert(), extend()
#l = []
#l = list()
#both are empty list
```

```python
#l = []
l = list()
n = int(input("Enter n = ")) #to decide how many values we want to append
for i in range(n):
    x = int(input("Enter x = "))
    l.append(x)
print(l)
```

```python
print("The maximum value = ", max(l))
print("The minimum value = ", min(l))
print("The sum of value = ", sum(l))
l.sort()
print("The sorted values in ascending order = ", l)
l.reverse()
print("The sorted values in descending order = ", l)
```

```python
#insert()
a = ['Ant', 'Apple', 'Ball', 'Orange', 'Zebra']
a.insert(1, 'Dog')
print(a)
```

```python
#extend()
a = ['Apple', 'Ball', 'Cat']
b = [1, 2, 3]
a.extend(b)
print(a)
```

```python
#CRUD operation (Creating, Read, Updating, Deleting)
```

```python
#Update
a = ['Ant', 'Apple', 'Apple', 'Ball', 'Orange', 'Zebra']
a[0] = 'ant' #list are mutable and dynamic
print(a)
```

```python
a[0:3] = ['aNt', 'apple', 'apple']
print(a)
```

```python
#Delete
a = ['aNt', 'apple', 'apple', 'Ball', 'Orange', 'Zebra']
del a[0]
print(a)
```

```python
a = ['aNt', 'apple', 'apple', 'Ball', 'Orange', 'Zebra']
del a[0:3]
print(a)
```

```python
#remove()
a = ['aNt', 'apple', 'apple', 'Ball', 'Orange', 'Zebra']
a.remove('Ball') #using remove() we can use the exact value
print(a)
```

```python
#pop()
a = ['aNt', 'apple', 'apple', 'Ball', 'Orange', 'Zebra']
a.pop(3) #using pop() we have to use index
print(a)
```

```python
#index()
a = ['aNt', 'apple', 'apple', 'Ball', 'Orange', 'Zebra']
a.index('Ball')
```
