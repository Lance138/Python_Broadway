```python
# Creating class and object in Python

class Point:   # naming convention
    def hello(self):    # method
        print("Hello")

# creating objects
point1 = Point()
point2 = Point()

# calling the method
point1.hello()
point2.hello()
```

```python
#self is a parameter. 
#It is a reference to the current instance of the class. 
#It has to be the first parameter of any function of the class.
```

```python
class Hello:    #class
    def hello(self):    #method
        print("Hello World")

obj = Hello()    #object
obj.hello()
```

```python
#static method
class Hello:    #class
    @staticmethod
    def hello():    #method
        print("Hello World")

obj = Hello()    #object
obj.hello()
```

```python
class Cal:
    def area(self, l, b):
        a = l * b
        print(a)
        
    def volume(self, l, b, h):
        v = l * b * h
        print(v)
        
l = 10
b = 5
h = 2
obj = Cal()
obj.area(l, b)
obj.volume(l, b, h)
```

```python
#__init__ method
#dundermethods (double underscore methods)

class Cal:
    def __init__(self, l, b, h):
        self.l = l #can write self.l or self.length any
        self.b = b
        self.h = h
        
    def area(self):
        a = self.l * self.b
        print(a)
        
    def volume(self):
        v = self.l * self.b * self.h
        print(v)
        
l = 10
b = 5
h = 2
obj = Cal(l, b, h) #when object class is initialized, the above runs
obj.area()
obj.volume()
```

```python
#Using staticmethod on area to use a specific value. The volume remains as it is

class Cal:
    def __init__(self, l, b, h):
        self.l = l #can write self.l or self.length any
        self.b = b
        self.h = h
    @staticmethod    
    def area(l, b):
        a = l * b
        print(a)
        
    def volume(self):
        v = self.l * self.b * self.h
        print(v)
        
l = 10
b = 5
h = 2
obj = Cal(l, b, h) #when object class is initialized, the above runs
obj.area(5, 3)
obj.volume()
```

```python
#
class Area:
    def __init__(self, l, b):
        self.l = l
        self.b = b
        
    def area(self):
        a = self.l * self.b
        print(a)

class Volume:
    def __init__(self, l, b, h):
        self.l = l
        self.b = b
        self.h = h
        
    def volume(self):
        a = self.l * self.b * self.h
        print(a)
l = 10
b = 5
h = 2
obj1 = Area(l, b)
obj2 = Volume(l, b, h)
obj1.area()
obj2.volume()
```

```python
#make blocks of codes first before filling them in to make life easier
#when making blocks of codes, make sure there are no errors when running the code
class Area:
    def __init__(self, l, b):
        self.l = l
        self.b = b
    def area(self):
        a = self.l * self.b
        return a

class Volume:
    def __init__(self, l, b, h):
        self.l = l
        self.b = b
        self.h = h
    def volume(self):
        v = self.l * self.b * self.h
        return v

obj1 = Area(l, b)
obj2 = Volume(l, b, h)
ar = obj1.area()
vl = obj2.volume()
print("The area is = ", ar)
print("The volume is = ", vl)
```
