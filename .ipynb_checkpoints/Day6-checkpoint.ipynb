```python
for i in range(10):
    if i == 5:
        break #Here, if i value = 5, loop breaks (terminates)
    else:
        print(i)
```

```python
a = "Hello world. I am Python"
for i in a:
    if i == ".":
        break
    else:
        print(i, end = "") #end = "" shows output on a single line
```

```python
for i in range (10):
    if i == 5:
        continue #continue is used here to skip the value and continue the loop after the skipped parameter( basically skip "that" and  continue)
    else:
        print(i, end = "")
```

```python
#for else loop
#when for and else are on the same indent. IF-else is when if and else statements are on same indent
#Write a program to detect if the number is prime or composite

n = int(input("Enter n ="))
for i in range (2, n):
    if n%i == 0:
        print("The number is composite")
        break
else:
    print("The number is prime")
```

```python
#while
#while <condition>:
#    <operations>
```

```python
i = 0 #initialization
while i <= 5: #declaring while loop with condition
    print(i, "Hello World")
    i = i + 1 #in while, unlike for loop we have to increment ourselves
```

```python
i = 1
while i <= 5:
    print(i, "Hello World")
    i = i + 2
```

```python
#write a program to make a multiplication table using While loop
i = 1
x = int(input("Enter x = "))
while i <= 10:
    print(x, '*', i, "=", i*x)
    i = i + 1
```

```python
i = 0
s = 0
n = int(input("Enter n = "))
r = float(input("Enter r = "))
while i<n:
    x = int(input("Enter x = "))
    s = (s + x)*(1+r/100)
    i = i + 1

    print(s)
            
```

```python
#Wap to take the name and phone number of the user and concatenate them.
i = 0
info = ""
n = int(input("Enter n ="))
while i<n: #range(n)
    name = input("Enter name = ")
    phone = input("Enter phone = ")
    info = info + name + " " + phone + "\n"
    i = i + 1
print(info)
```
