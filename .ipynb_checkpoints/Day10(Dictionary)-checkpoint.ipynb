```python
d = dict()
print(d)
```

```python
#d = {<key>:<value>, ,<key>:<value>}
```

```python
#key should always be unique and never be the same
d = {'a':'apple', 'b':'ball', 'c':'cat', 'd':'dog'}
print(d)
```

```python
#indexing
print(d['a'])
print(d['b'])
print(d['c'])
print(d['d'])
```

```python
#Create
d = {}
d['a'] = 'apple'
d['b'] = 'ball'
d['c'] = 'cat'
d['d'] = 'dog'
print(d)
```

```python
#This only prints keys
for i in d:
    print(i)
```

```python
#This only prints values
for i in d.values():
    print(i)
```

```python
#This prints items (keys and values)
for i in d.items():
    print(i)
```

```python
#add data
d = {'a':'apple', 'b':'ball', 'c':'cat', 'd':'dog'}
d['e'] = 'elephant'
d
```

```python
#update
d = {'a':'apple', 'b':'ball', 'c':'cat', 'd':'dog'}
d['a'] = 'Apple' #updating with 'A' over 'a' in apple as example
d
```

```python
#delete
d = {'a':'apple', 'b':'ball', 'c':'cat', 'd':'dog'}
del d['a']
print(d)
```

```python
#create list inside dict
data1 = {}
n = int(input("Enter no. of entries = "))
for i in range(n):
    name = input("enter name = ")
    ntc_phone = int(input("enter ntc phone = "))
    ncell_phone = int(input("enter ncell phone = "))
    data1[name] = [ntc_phone, ncell_phone]

data1
```

```python
#dict inside list
data = []
n = int(input("enter n = "))
for i in range(n):
    name = input("Enter name = ")
    ntc_phone = int(input("Enter ntc phone = "))
    ncell_phone = int(input("Enter ncell phone = "))
    x = {'Name':name, 'Ntc':ntc_phone, 'Ncell':ncell_phone}
    data.append(x)

data
```

```python
print(data[0]['Name'])
```
