```python
# WAP to take the name and phone number of the user and concatenate them.
a = input("Enter name = ")
b = (input("Enter phone number = "))
c = a +" "+ b
print(c)
```

```python
# WAP to take marks of multiple students having multiple subjects and calculate total, per, grade and concate them
a = int(input("Enter number of students = "))
b = int(input("Enter number of subjects = "))
for i in range (a):
    name = input("\nEnter name of the student = ")
    total = 0
    
    for j in range (b):
        subject = input("\nEnter name of the subject =")
        marks = int(input("Enter marks in " + subject + " =" ))
        total += marks

    percent = total / b

    if percent >= 90:
        grade = "A"
    elif percent >= 70:
        grade = "B"
    elif percent >= 60:
        grade = "C"
    elif percent >= 40:
        grade = "D"
    else:
        grade ="Fail"

    result = name + " Total = " + str(total) + " Percentage = " + str(percent) + " Grade = " + grade
    print(result)
        
            
        
```

```python
# WAP to detect if the number is prime or composite

n = int(input("Enter a number = "))
if n <= 1:
    print("It is neither Prime nor a Composite number")
else:
    is_prime = True
    for i in range (2, n):
        if n % i == 0:
            is_prime = False
            break
            
    if is_prime:
        print("It is a Prime number")
    else:
        print("It is a Composite number")
        
```
