Rules
Variables should not start with numbers (0-9)
No special characters ((!@#$%^&*()-+=))
No space between variables

syntax error = grammar error in code language

```python
a = 10
b = 3
print(a + b)
print(a - b)
print(a * b)
print(a / b)

print(a%b)
print(a**b) #power
print(a//b)
```

```python
a = True #True = 1
b = False #False = 0
print(a + b)
```

```python
#Type casting or data type conversion
# int(), float(), str(), bool()
a = 10 #int
b = float(a) #conversion to float
print(b)
```

```python
a = 10.5 #float
b = int(a) #conversion to int
print(b)
```

```python
a = 10
b = str(a) # output is string even though it looks like int
print(b)
```

```python
a = 10
b = 0 
c = bool(a) # every value is converted to True unless it is 0. If value is 0 then it is converted to False
d = bool(b) # every value is converted to True unless it is 0. If value is 0 then it is converted to False
print(c)
print(d)
```

```python
a = "Hello"
b = ""
a = bool(a) #Non-empty string is considered True
b = bool(b) #Empty string is considered False
print(a)
print(b)
```
