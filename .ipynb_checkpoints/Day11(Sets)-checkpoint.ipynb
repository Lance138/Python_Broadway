```python
s = set()
type(s)
```

```python
#does not take duplicate values
s = {"Apple", "Ball", "Cat","Dog", "Apple"}
print(s)
```

```python
#we cannot index in set
print(s[0])
```

```python
#this is unordered
s = {"Apple", "Ball", "Cat","Dog", "Apple"}
print(s)
```

```python
#we don't use loops too much in sets
s = {"Apple", "Ball", "Cat","Dog", "Apple"}
for i in s:
    print(s)
```

```python
#in set no +, *
#we cannot take two sets and concat them
#we can add new values in the set
```

```python
#in set .add() .update() .remove() work
```

```python
s = set()
n = int(input("Enter n = "))
for i in range(n):
    x = int(input("Enter x = "))
    s.add(x)
print(s)
```

```python
#.update() on set
a = {1,2,3}
b = {4,5,6}
a.update(b)
print(a)
```

```python
#.remove()
a = {1,2,3,4,5,6}
a.remove(2)
print(a)
```

```python
# only immutable datatype inside set
# int, float, bool,tuple, inside set possible
# list, dict, set, inside set impossible
# set can be in any like list, dict but set cannot be in set
```

```python
s = {(1,2,3),(4,5,6)}
s
```

```python
#set in math
# union()
# intersection()
# difference()
```

```python
# .union() operation. Duplicate values are not inserted
a = {1,3,5,7,9,10}
b = {2,4,6,8,10}
c = a.union(b)
c
```

```python
# .intersection() operation (common elements in a and b)
a = {1,3,5,7,9,10}
b = {2,4,6,8,10}
c = a.intersection(b)
c
```

```python
# .difference() operation (all elements in set a that are not in set b)
a = {1,3,5,7,9,10}
b = {2,4,6,8,10}
c = a.difference(b)
c
```

```python
# all elements in set b that are not in set a)
b.difference(a)
```

```python
#Common employees in both companies
apple = {'John', 'Nabin', 'Steve', 'Sahil', 'Krish'}
ms = {'Tanka', 'Bill', 'Sahil', 'Sabina', 'Rita', 'Krish'}
apple.intersection(ms)
```

```python
#IF the companies merge
print(apple.union(ms))
```

```python
#Finding employees only working in apple and not microsoft
apple - ms
```

```python
#Finding employees only working in microsoft and not apple
ms - apple
```
