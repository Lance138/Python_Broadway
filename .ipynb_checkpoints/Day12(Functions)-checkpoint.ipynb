```python
#local variable
def area():
    l = int(input("Enter l = ")) #local variable (cannot be used outside function)
    b = int(input("Enter b = ")) #local variable (cannot be used outside function)
    a = l*b                      #local variable (cannot be used outside function)
    print(a)

area()
```

```python
#Global variable 1
l = int(input("Enter l = ")) #Global variable (It is outside of the function indentation)
b = int(input("Enter b = ")) #Global variable (It is outside of the function indentation)
def area():
    a = l*b
    print(a)
area()
```

```python
#Global variable 2 (Function with no arguments and no return type)
def area():
    global l,b                   #we are specifying the global variables inside function indentation
    l = int(input("Enter l = ")) #Global variable
    b = int(input("Enter b = ")) #Global variable
    a = l * b                    #Local variable
    print(a)
area()
print(l,b)
```

```python
#parameter passing (passing value to a function)
#Example of parameter passing in a user defined function
def area(x,y): #parameter
    a = x * y #local variable
    print(a)

l = int(input("Enter l = ")) #global variable
b = int(input("Enter b = ")) #global variable
area(l, b) #l and b is argument
```

```python
#Function with argument and non return type
def info(name, age, add):
    data = f"Hello I am {name}. I am {age} years old and I am from {add}."
    print(data)
name = input("Enter name = ")
age = int(input("Enter age = "))
add = input("Enter address = ")
info(name, age, add)
```

```python
#list as argument
#Function with argument and non return type
def info(x):
    name, age, add = x
    data = f"Hello I am {name}. I am {age} years old and I am from {add}."
    print(data)
name = input("Enter name = ")
age = int(input("Enter age = "))
add = input("Enter address = ")
x = [name, age, add]
info(x)
```

```python
def area(l, b):
    a = l * b
    return a
    
l = int(input("Enter l = "))
b = int(input("Enter b = "))
h = int(input("Enter h = "))
ar = area(l, b)
v = ar * h #volume
print ("The area = ", ar)
print("The volume = ", v)
```

```python
def cal(l,b,h):
    a = l*b
    v = a*h
    return a,v
l = int(input("Enter l = "))
b = int(input("Enter b = "))
h = int(input("Enter h = "))

area, volume = cal(l,b,h) 
print("The area is = ", area)
print("The volume is = ", volume)
```

```python
#WAP to create a calculator using return type function

def cal(a,b):
    add = a + b
    sub = a - b
    mult = a * b
    div = a / b if b != 0 else "Undefined"
    return add, sub, mult, div
a = int(input("Enter a = "))
b = int(input("Enter b = "))

add, sub, mult, div = cal(a, b) 
print("The addition is = ", add)
print("The subtraction is = ", sub)
print("The multiplication is = ", mult)
print("The division is = ", div)
```

```python
#WAP to create a calculator using return type function
#Extra with a selection option for specific operation
def cal(a, b, op):
    if op == 1:
        return a + b
    elif op == 2:
        return a - b
    elif op == 3:
        return a * b
    elif op == 4:
        if b == 0:
            return "Division not possible"
        return a / b
    else:
        return "Invalid choice"


a = int(input("Enter a = "))
b = int(input("Enter b = "))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice = "))

result = cal(a, b, choice)

print("Result =", result)
```

```python
#WAP to create a billing system with all total price using return type function

def bill(price, qty):
    total = price * qty
    return total

item = input("Enter item name = ")
price = int(input("Enter price =  "))
qty = int(input("Enter quantity = "))
total_price = bill(price , qty)
print("Item = ", item)
print("Total bill = ", total_price) 
```

```python
#WAP to create a billing system with all total price using return type function

def bill():
    total = 0
    n = int(input("Enter number of items = "))

    for i in range(n):
        item = input("Enter item name = ")
        price = float(input("Enter price = "))
        quantity = int(input("Enter quantity = "))
        total += price * quantity
        
    return total
billing = bill()
print("Total = ", billing)
```

```python
#WAP to give an example of function inside function

def cal(a, b):
    def add(x, y):
        return x + y

    def sub(x, y):
        return x - y

    def mult(x, y):
        return x * y

    def div(x, y):
        return x / y if y != 0 else "Undefined"

    return add(a,b), sub(a, b), mult(a, b), div(a, b)

print(cal(10, 5))
```

```python
#WAP to take inputs from user and create matrix
matrix = []
def get_row():
    for i in range(3):
        row = []
        for j in range(3):
            value = int(input(f"Enter value for row {i+1}, column {j+1}: "))
            row.append(value)
        matrix.append(row)
        
get_row()

for row in matrix:
    print(row)
#print(matrix)
```
