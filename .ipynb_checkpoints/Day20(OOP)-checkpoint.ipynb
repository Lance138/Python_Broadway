```python
# __str__method
# This method returns the string representation of the object. This method is called when print() or str() function is invoked on an object.
```

```python
# str method
# This cannot give other data types besides string
class Info:
    def __init__(self):
        self.name = "Ram"
        self.age = 34
        self.add = "Kathmandu"
        
    def __str__(self):
        return self.name

obj = Info()
print(obj)
print(type(obj))
print(type(obj.__str__()))
obj.__str__()
```

```python
#__repr__ method

class Info:
    def __init__(self):
        self.name = "Ram"
        self.age = 34
        self.add = "Kathmandu"
        
    def __repr__(self):
        return {'name':self.name, 'age':self.age, 'add':self.add}

obj = Info()
print(type(obj))
print(obj.__repr__())
print(type(obj.__repr__()))
```

```python
# using __repr__ method practice

class Info:
    def __init__(self, name, age, add):
        self.name = name
        self.age = age
        self.add = add

    def __repr__(self, name, age, add):
        return {'name':self.name, 'age':self.age, 'add':self.add}

name = input("Enter name = ")
age = int(input("Enter age = "))
add = input("Enter address = ")
obj = Info(name, age, add)
print(obj.__repr__(name, age, add))
```

```python
# Polymorphism in Python
class Parrot:
    def fly(self):
        print("Parrot can fly")
    def swim(self):
        print("Parrot cannot swim")

class Penguin:
    def fly(self):
        print("Penguin can't fly")
    def swim(self):
        print("Penguin can swim")

# Common interface
def flying_test(bird):
    bird.fly()

# Swimming Test
def swimming_test(bird):
    bird.swim()

# instantiate objects
blu = Parrot()
peggy = Penguin()

# passing the object
flying_test(blu)
flying_test(peggy)

# Swimming Test
swimming_test(blu)
swimming_test(peggy)
```

```python
# Operator Overloading
#Error showns when running this code

# class Point:
#     def __init__(self, x, y):
#         self.x = x
#         self.y = y

# p1 = Point(2,3)
# p2 = Point(-1,2)
# p1 + p2
```

```python
# Operator Overloading

class Point:
    def __init__(self, x = 0, y = 0):
        self.x = x
        self.y = y

    def __str__(self):
        return "({0},{1}".format(self.x,self.y)

    def __add__(self,other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x,y)

p1 = Point(2,3)
p2 = Point(-1,2)
print(p1+p2)
```

```python
# # Operator Overloading Special Functions in Python

# Operator
# Expression
# Internally

# Addition
# p1 + p2
# p1__add__(p2)

# Subtraction
# p1 - p2
# p1__sub__(p2)

# Multiplication
# p1 * p2
# p1__mul__(p2)

# Power
# p1 ** p2
# p1__pow__(p2)

# Division
# p1 / p2
# p1__truediv__(p2)

# Floor Division
# p1 // p2
# p1__floordiv__(p2)

# Remainder (modulo)
# p1 % p2
# p1__mod__(p2)
```

```python
# Operator Overloading
# This shows what part of code runs first and what each code block does

class Point:
    def __init__(self, x, y):       # Initializes object attributes when a new object is created
        self.x = x
        self.y = y
        print("This is init method", self.x, self.y)

    def __str__(self):            # Defines how the object is displayed when printed
        print("This is str method", self.x, self.y)
        return f"{self.x}, {self.y}"

# Defines behavior of + operator for Point objects  
# Returns a new Point object with added values
    def __add__(self,other):  
        x = self.x + other.x
        y = self.y + other.y
        print("This is add method", x, y)
        return Point(x,y)

p1 = Point(2,3)
p2 = Point(-1,2)
print(p1+p2)      # __add__ returns a new Point object; this triggers __init__, and print() calls __str__ to display it
```
