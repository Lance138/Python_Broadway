```python
#billing system
bill = ""
grand_total = 0
n = int(input("Enter n = "))
for i in range(n):
    product = input("Enter the name = ")
    price = int(input("Enter price = "))
    qty = float(input("Enter quantity = "))
    total = price * qty
    data = f"{product}\t{price}\t{ qty}\t{total}\n"
    bill = bill + data
    grand_total = grand_total + total
    
print(bill)
print(grand_total)

bill = bill + "Total = " + str(grand_total)
f = open("bill.txt", 'w')
f.write("Product \tPrice \tQuantity \ttotal\n")
f.write(bill)
f.close()
```

```python
#split()
#Seperates the data
#Here, it splits the data by \n which is next line
data.split("\n")
```

```python
#Here, it splits the data by \t which is tab i.e. indentation
data.split("\t")
```

```python
#Here data splits by \n and it takes all values from index 1 to -1 not including index -1
data1 = data.split("\n")[1:-1]
data1
```

```python
#Converting to array splitting by \t which is tab(indentation)
for i in data1:
    print(i.split("\t"))
```

```python
#converting the data to integer
for i in data1:
    x = i.split("\t")
    x[1] = float(x[1])
    x[2] = float(x[2])
    x[3] = float(x[3])
    print(x)
```
