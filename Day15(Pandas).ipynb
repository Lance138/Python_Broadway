```python
#filter
#single brackets only work when there is only one value
df['Age']
```

```python
#filter by age number
df['Age'] > 34
```

```python
#filter
#This only shows data with age over 34 years old
df[df["Age"]>34]
```

```python
# showing data with age over 40 and has an education level equal to PhD
df[(df['Age'] > 40) & (df["Education Level"] == "PhD")]
```

```python
# showing data with age over 40 or has an education level equal to PhD
df[(df['Age'] > 40) | (df["Education Level"] == "PhD")]
```

```python
# How to convert list into data frame
data = [["Ram", 34, "Kathmandu"],
       ["Shyam",45,"Bhaktapur"],
       ["Hari",34,"Lalitpur"]]
col = ["Name","Age","Address"]
df = pd.DataFrame(data, columns = col)
df
```

```python
# How to convert dict into data frame
data1 = {'Name':["Ram","Shyam","Hari"],
       'Age':[23, 45, 65],
        'Address':["Kathmandu","Bhaktapur","Lalitpur"]}
df1 = pd.DataFrame(data1)
df1
```

```python
# Concat two dataframes
data1 = {'Name':["Ram","Shyam","Hari"],
       'Age':[23, 45, 65],
        'Address':["Kathmandu","Bhaktapur","Lalitpur"]}
df1 = pd.DataFrame(data1)

data2 = {'Name':["Ram","Shyam","Hari"],
       'Age':[23, 45, 65],
        'Address':["Kathmandu","Bhaktapur","Lalitpur"]}
df2 = pd.DataFrame(data2)
df = pd.concat([df1,df2])
df
```
