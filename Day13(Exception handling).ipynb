```python
#Exception handling: inputting string on an int value input
try:
    x = int(input("Enter x = "))
except:
    print("There is some error")
```

```python
#Exception handing: division error (zero division, str value, etc)
try:
    a = int(input("Enter a = "))
    b = int(input("Enter b = "))
    c = a / b
    print(c)
    
except:
    print("There is some error")
```

```python
#Exception handing: try, except, else, finally
#finally block runs everytime
try:
    a = int(input("Enter a = "))
    b = int(input("Enter b = "))
    c = a / b
    
except ValueError:
    print("The value of a and b should always be number")
except ZeroDivisionError:
    print("Cannot divide by Zero")
else:
    print(c)
finally:
    print("This is finally block")
```

```python
# raise : raises errors based on user request
raise ValueError
```
